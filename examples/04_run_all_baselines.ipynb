# Run NL2T and all seven baselines on a small world.

```
from nl2t.env.world import default_world
from nl2t.sim.engine import simulate
from nl2t.baselines import BASELINES
from nl2t.baselines.nl2t_protocol import NL2TProtocol

world = default_world(seed=0, n_nodes=20, anchor_frac=0.3)
results = {'NL2T': simulate(NL2TProtocol(), world, duration_h=0.2, seed=0)}
for name, cls in BASELINES.items():
    results[name] = simulate(cls(), world, duration_h=0.2, seed=0)
for name, r in results.items():
    print(f'{name:12s}  PDR={r.pdr:.3f}  loc_err={r.mean_loc_error:.2f}m  E/bit={r.energy_per_bit_nj:6.0f}nJ')
```
