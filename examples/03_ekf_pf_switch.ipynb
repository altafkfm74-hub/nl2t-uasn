# Adaptive EKF↔PF switching

Eq. (26) of NL2T.

```
from nl2t.core import coupling, tracker
for Sp, Sm, Er in [(0.0, 0.0, 50.0), (10.0, 0.0, 50.0), (0.0, 0.0, 3.0)]:
    f = coupling.adaptive_filter_energy_coupled(Sp, Sm, h=5.62, E_res=Er, E_th_pf=10.0)
    print(f'S+={Sp:4.1f}  S-={Sm:4.1f}  E={Er:5.1f}J  ->  {f}')
```
